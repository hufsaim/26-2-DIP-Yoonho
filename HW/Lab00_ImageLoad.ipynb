- Image file의 metadata를 확인하기

```python
from PIL import Image
from PIL.ExifTags import TAGS
```

```python
I = Image.open('IMG_6562.jpeg')
exifdata = I.getexif()
for tag_id in exifdata:
    tag = TAGS.get(tag_id, tag_id)
    data = exifdata.get(tag_id)
    print(f"{tag:25}: {data}")
```
